رگرسیون :پیدا کردن ارتباط بین متغیرهای ورودی و خروجی
متعیر ورودی= متغیر مستقل
متغیر خروجی=متغیر وابسته

انواع رگرسیون:1.رگرسیون تک متغیره (ساده)
   1.1 رگرسیون خطی ساده
   1.2رگرسیون غیر خطی ساده

2.رگرسیون چند متغیره
   2.1رگرسیون خطی چندگانه
   2.2 رگرسیون غیر خطی چندگانه

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# data
# y=2*x-1
x= np.array([1,2,3,4]).reshape(-1,1)
y=np.array([1,3,5,7])
```

```python
x.shape
```

```python
x
```

```python
# plot
plt.plot(x,y,'*')
plt.show()
```

```python
# model
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x,y)
```

```python
# y=m*x+c
# m=coef ایکس ضریب cc _,c=intercept_
m=model.coef_[0]
c=model.intercept_
print(m,c)
```

```python
# رگرسیون ساده تک متغیره
plt.plot(x,y,'*')
plt.plot(x,model.predict(x))
plt.show()
```

```python
result=model.predict([[5]])
result
```

```python
years_of_experience = np.array([0,1,2,3,4,6,8,10]).reshape(-1,1)
salaries= np.array([30000,40000,55000,60000,70000,80000,85000,87000])
plt.plot(years_of_experience,salaries,"*")
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor
model=DecisionTreeRegressor(max_depth=3)
model.fit(years_of_experience,salaries)
y_predict=model.predict(years_of_experience)
plt.plot(years_of_experience,salaries,'*')
plt.plot(years_of_experience,y_predict)
plt.show()
```
